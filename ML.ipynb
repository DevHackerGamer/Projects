```python
import numpy as np
import pandas as pd
import sklearn
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
data = pd.read_csv("traindata.txt").iloc[:, :-1]
labels = pd.read_csv("trainlabels.txt")
```

```python
#replaces NaN with 1 in the DataFrame - results in higher accuracy
#data = data.fillna(1)

#drops NaN values
#data = data.dropna(axis=0)

data_arr = data.to_numpy()

#Change NaN values to 0 - will need to experiment with this to see if there's a better way of handling it
data_arr = np.nan_to_num(data_arr)
labels_arr = labels.to_numpy().ravel() #to flatten the array, otherwise would come out as a 250x1 array
```

```python
data_train, data_test, labels_train, labels_test = train_test_split(data_arr, labels_arr, test_size=0.2, random_state=42)
```

```python
#Save training data
pd.DataFrame(data_train).to_csv("data_train.csv", index=False)

#Save training labels
pd.DataFrame(labels_train).to_csv("labels_train.csv", index=False)

# Save test data
pd.DataFrame(data_test).to_csv("data_test.csv", index=False)

# Save test labels
pd.DataFrame(labels_test).to_csv("labels_test.csv", index=False)
```

```python
model = GaussianNB(var_smoothing = 1e-09)
model.fit(data_train, labels_train)

labels_pred = model.predict(data_test)

accuracy = accuracy_score(labels_test, labels_pred)
print(f"Accuracy: {accuracy:.4f}")
```
