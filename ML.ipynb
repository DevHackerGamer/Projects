```python
import numpy as np
import pandas as pd
import sklearn
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import f_classif

data = pd.read_csv("traindata.txt").iloc[:, :-1]
print(data.shape)

#converting the values from strings to numerical values
for column in data.columns:
    data[column] = pd.to_numeric(data[column], errors='coerce')

# replaces NaN with 1 in the DataFrame - results in higher accuracy
# data = data.fillna(1)

# drops NaN values
# data = data.dropna(axis=0)

# going to try to fill NaN values with the mean of that column, then try with the mean of that row

for column in data.columns:
    data[column] = data[column].fillna(data[column].mean())

#num_rows = data_train_new.shape[0]

#for i in range(num_rows):
#    row = data[i]
#    mean = np.nanmean(row)
    # replace NaNs in the row with the mean of that row
#    nan_index = np.isnan(row)
#    row[nan_index] = mean
#    data[i] = row



labels = pd.read_csv("trainlabels.txt")

# convert data and labels to arrays for train_test_split
data_arr = data.to_numpy()
labels_arr = labels.to_numpy().ravel()  # flatten labels array

data_train, data_test, labels_train, labels_test = train_test_split(data_arr, labels_arr, test_size=0.2, random_state=42)

# using SelectKBest
selector = SelectKBest(score_func=f_classif, k=17)

# apply selector to my data
selector.fit(data_train, labels_train)

# now both training and testing data work with top 17 features
data_train_new = selector.transform(data_train)
data_test_new = selector.transform(data_test)

# replaces NaN with 1 in the DataFrame - results in higher accuracy
# data = data.fillna(1)

# drops NaN values
# data = data.dropna(axis=0)

data_arr = data_train_new  # already numpy array, no need

# Change NaN values to 0 - will need to experiment with this to see if there's a better way of handling it
#data_arr = np.nan_to_num(data_arr)

# to flatten the array, otherwise would come out as a 250x1 array
labels_arr = labels_train.ravel()  # already numpy array and raveled

```

```python
#Save training data
pd.DataFrame(data_train).to_csv("data_train.csv", index=False)

#Save training labels
pd.DataFrame(labels_train).to_csv("labels_train.csv", index=False)

# Save test data
pd.DataFrame(data_test).to_csv("data_test.csv", index=False)

# Save test labels
pd.DataFrame(labels_test).to_csv("labels_test.csv", index=False)
```

```python
model = GaussianNB(var_smoothing = 1e-09)

model.fit(data_train_new, labels_train)

labels_pred = model.predict(data_test_new)

accuracy = accuracy_score(labels_test, labels_pred)
print(f"Accuracy: {accuracy:.4f}")
```
